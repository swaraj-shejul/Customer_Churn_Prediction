# Model 2 - Multi layer Perceptron

```python
import pandas as pd
import numpy as np
import tensorflow as tf
import keras
from keras.models import Sequential
from keras.layers import Dense
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import random


tf.random.set_seed(42)
np.random.seed(42)
random.seed(42)

```

```python
customer_data = pd.read_excel('../Datasets/cleaned_customers_churn_data.xlsx')
customer_data.head()
```

```python
customer_data.shape
```

```python
x = customer_data.drop(columns=['Exited'])
y = customer_data['Exited']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
# Designing the MLP model
MLP_model = Sequential()
MLP_model.add(Dense(input_dim=11, units=22, activation='relu', kernel_initializer='uniform'))
MLP_model.add(Dense(units=1, activation='sigmoid', kernel_initializer='uniform'))
# compiling the model
MLP_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
# fitting the model
MLP_model.fit(x_train,y_train, batch_size=32, epochs=100)
# predicting the trained model
y_pred_train = MLP_model.predict(x_train)
y_pred_train = (y_pred_train > 0.5)

# predicting the test set results
y_pred_test = MLP_model.predict(x_test)
y_pred_test = (y_pred_test > 0.5)

# Evaluating the model
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
print("Training accuracy : ", accuracy_score(y_train, y_pred_train))
print("Test accuracy : ", accuracy_score(y_test, y_pred_test))
```

- Implemented a Multi-Layer Perceptron (MLP) with 1 hidden layer containing 22 neurons (2 × input features).
- Training Accuracy = 85.86%
- Testing Accuracy = 86.10%
- Compared to the Perceptron model, the MLP achieved better overall performance across all evaluation metrics.


```python
confusion_matrix(y_test, y_pred_test)
```

- Correctly identified 186 churn customers, compared to only 81 in the Perceptron model.
- False Negatives decreased from 312 → 207, meaning fewer churn customers were missed.
- The model became significantly better at identifying customers likely to leave the bank.

```python
print(classification_report(y_test, y_pred_test))
```

```python

from sklearn.metrics import roc_auc_score


print("ROC AUC Score:", roc_auc_score(y_test, y_pred_test))
```

- Recall improved from 21% to 47%, meaning the model can now identify nearly half of the customers who will churn.
- Precision improved from 56% to 72%, reducing false churn predictions.
- F1 Score increased substantially, indicating a better balance between Precision and Recall.
- ROC-AUC improved from 0.58 to 0.71, showing a much stronger ability to separate churn and non-churn customers.

# Conclusion 

Adding a hidden layer significantly improved the model's ability to learn non-linear relationships in customer behavior. The MLP outperformed the Perceptron across Accuracy, Precision, Recall, F1-Score, and ROC-AUC while maintaining good generalization. Since performance improved considerably with a single hidden layer, experimenting with a deeper neural network (additional hidden layers) may further improve results. However, increasing network depth will also increase model complexity and training time, so performance gains should be validated against potential overfitting.
